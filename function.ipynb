```python
def get_marks():
    tamil = int(input("Enter Tamil Marks : "))
    english = int(input("Enter English Marks : "))
    maths = int(input("Enter Maths Marks : "))
    science = int(input("Enter Science Marks : "))
    social = int(input("Enter Social Marks : "))

    return tamil, english, maths, science, social


def calculate_total(tamil, english, maths, science, social):
    total = tamil + english + maths + science + social
    return total


def calculate_average(total):
    average = total / 5
    return average


def find_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "Fail"


def display_result(total, average, grade):
    print("\n===== Student Result =====")
    print("Total Marks :", total)
    print("Average Marks :", average)
    print("Grade :", grade)



tamil, english, maths, science, social = get_marks()

total = calculate_total(
    tamil, english, maths, science, social
)

average = calculate_average(total)

grade = find_grade(average)

display_result(total, average, grade)
```

```python
products = {
    101: ("Laptop", 55000),
    102: ("Mouse", 600),
    103: ("Keyboard", 1200),
    104: ("Monitor", 12000),
    105: ("Headphone", 2500),
    106: ("Pendrive", 700),
    107: ("Webcam", 3000),
    108: ("Printer", 9000)
}

cart = {}

def display_menu():
    print("\n===== TECHLIFE MART =====")
    print("1. Display Products")
    print("2. Add Product")
    print("3. Remove Product")
    print("4. Update Quantity")
    print("5. Search Product")
    print("6. View Shopping Cart")
    print("7. Calculate Bill")
    print("8. Generate Invoice")
    print("9. Exit")

def display_products():
    print("\n===== AVAILABLE PRODUCTS =====")

    for product_id, details in products.items():
        name = details[0]
        price = details[1]

        print(
            "ID:", product_id,
            "| Name:", name,
            "| Price:", price
        )
def add_product():

    display_products()

    product_id = int(input("Enter Product ID : "))

    if product_id not in products:
        print("Invalid Product ID")
        return

    quantity = int(input("Enter Quantity : "))

    if quantity <= 0:
        print("Quantity must be greater than zero")
        return

    name, price = products[product_id]

    if product_id in cart:

        cart[product_id]["quantity"] += quantity

        print("Product quantity increased")

    else:

        cart[product_id] = {
            "name": name,
            "price": price,
            "quantity": quantity
        }

        print("Product added successfully")

def remove_product():

    if not cart:
        print("Shopping Cart is Empty")
        return

    product_id = int(
        input("Enter Product ID to remove : ")
    )

    if product_id in cart:

        removed = cart.pop(product_id)

        print(
            removed["name"],
            "removed successfully"
        )

    else:
        print("Product Not Found")

def update_quantity():

    if not cart:
        print("Shopping Cart is Empty")
        return

    product_id = int(
        input("Enter Product ID : ")
    )

    if product_id not in cart:
        print("Product Not Found")
        return

    quantity = int(
        input("Enter New Quantity : ")
    )

    if quantity <= 0:

        cart.pop(product_id)

        print("Product removed from cart")

    else:

        cart[product_id]["quantity"] = quantity

        print("Quantity updated successfully")

def search_product():

    print("\n1. Search by ID")
    print("2. Search by Name")

    choice = input("Enter Choice : ")

    if choice == "1":

        product_id = int(
            input("Enter Product ID : ")
        )

        if product_id in products:

            name, price = products[product_id]

            print("Product ID :", product_id)
            print("Product Name :", name)
            print("Price :", price)

        else:
            print("Product Not Found")

    elif choice == "2":

        search_name = input(
            "Enter Product Name : "
        ).lower()

        found = False

        for product_id, details in products.items():

            name = details[0]
            price = details[1]

            if search_name in name.lower():

                print("Product ID :", product_id)
                print("Product Name :", name)
                print("Price :", price)

                found = True

        if not found:
            print("Product Not Found")

    else:
        print("Invalid Choice")

def calculate_bill():

    if not cart:
        print("Shopping Cart is Empty")
        return

    subtotal = 0
    total_items = 0

    for details in cart.values():

        price = details["price"]
        quantity = details["quantity"]

        subtotal += price * quantity
        total_items += quantity


    if subtotal >= 20000:
        discount_percentage = 15

    elif subtotal >= 10000:
        discount_percentage = 10

    elif subtotal >= 5000:
        discount_percentage = 5

    else:
        discount_percentage = 0

    discount = (
        subtotal * discount_percentage / 100
    )

   

    gst = subtotal * 18 / 100

   

    grand_total = (
        subtotal
        - discount
        + gst
    )

    print("\n===== BILL =====")

    print("Total Items :", total_items)
    print("Subtotal :", subtotal)
    print("Discount :", discount)
    print("GST :", gst)
    print("Grand Total :", grand_total)

from datetime import datetime


def generate_invoice():

    if not cart:
        print("Shopping Cart is Empty")
        return

    customer_name = input(
        "Enter Customer Name : "
    )

    mobile = input(
        "Enter Mobile Number : "
    )

    cashier_name = input(
        "Enter Cashier Name : "
    )

    subtotal = 0
    total_items = 0

    for details in cart.values():

        price = details["price"]
        quantity = details["quantity"]

        subtotal += price * quantity
        total_items += quantity


    if subtotal >= 20000:
        discount_percentage = 15

    elif subtotal >= 10000:
        discount_percentage = 10

    elif subtotal >= 5000:
        discount_percentage = 5

    else:
        discount_percentage = 0

    discount = subtotal * discount_percentage / 100

    gst = subtotal * 18 / 100

    grand_total = (
        subtotal
        - discount
        + gst
    )

    print("\n")
    print("=" * 50)
    print("           TECHLIFE MART")
    print("=" * 50)

    print("Customer :", customer_name)
    print("Mobile   :", mobile)
    print("Cashier  :", cashier_name)

    print(
        "Date     :",
        datetime.now().strftime("%d-%m-%Y %H:%M:%S")
    )

    print("-" * 50)

    for product_id, details in cart.items():

        name = details["name"]
        price = details["price"]
        quantity = details["quantity"]

        total = price * quantity

        print(name)
        print(
            price,
            "×",
            quantity,
            "=",
            total
        )

    print("-" * 50)

    print("Total Items :", total_items)
    print("Subtotal :", subtotal)
    print(
        f"Discount ({discount_percentage}%) :",
        discount
    )
    print("GST (18%) :", gst)
    print("Grand Total :", grand_total)

    print("=" * 50)

    print("Thank You For Shopping With TechLife Mart")
    print("Visit Again!")

def main():

    while True:

        display_menu()

        choice = input(
            "Enter your choice : "
        )

        if choice == "1":

            display_products()

        elif choice == "2":

            add_product()

        elif choice == "3":

            remove_product()

        elif choice == "4":

            update_quantity()

        elif choice == "5":

            search_product()

        elif choice == "6":

            display_cart()

        elif choice == "7":

            calculate_bill()

        elif choice == "8":

            generate_invoice()

        elif choice == "9":

            print(
                "Thank you for using TechLife Mart!"
            )

            break

        else:

            print("Invalid Choice")

main()
```
